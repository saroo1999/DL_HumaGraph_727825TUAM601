# Model Evaluation
This notebook evaluates the trained Autoencoder + MLP model.

```python
import pandas as pd
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

X_test = pd.read_csv('../datasets/X_test.csv')
y_test = pd.read_csv('../datasets/y_test.csv')

encoder = tf.keras.models.load_model('../models/encoder.h5')
classifier = tf.keras.models.load_model('../models/best_model.h5')

encoded_X_test = encoder.predict(X_test)
y_pred = (classifier.predict(encoded_X_test) > 0.5).astype(int)
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d')
plt.title('Confusion Matrix')
plt.show()
```
